## EDA

### Histogram

```python


for column in numerical_columns:
    if column!='id':
        plt.figure(figsize=(6,4))
        sns.histplot(df[column], kde=True, bins=30)
        plt.title(f"Histogram of {column}")
        plt.xlabel(column)
        plt.ylabel("Frequency")
        plt.show()


```

```python
for column in numerical_columns:
    if column!='id':
       plt.figure(figsize=(6,4))
       sns.boxplot(y=df[column])
       plt.title(f"Box Plot of {column}")
       plt.xlabel(column)
       plt.show()
```

## Outlier Handling---> Caping/Winsorization Method

```python


columns_to_handle = [
    "price",
    "sqft_living",
    "sqft_lot",
    "sqft_above",
    "sqft_basement"
]

for column in columns_to_handle:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df[column] = df[column].clip(lower=lower, upper=upper)
```

```python
for column in numerical_columns:
    if column!='id':
       plt.figure(figsize=(6,4))
       sns.boxplot(y=df[column])
       plt.title(f"Box Plot of {column}")
       plt.xlabel(column)
       plt.show()
```


Outlier treatment was applied only to continuous numerical features. Identifier, categorical, location, and year-related columns were retained without modification because their values represent valid categories or attributes rather than continuous measurements.
