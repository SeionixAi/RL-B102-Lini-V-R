
## 1: What is Machine Learning?

Machine Learning is a branch of Artificial Intelligence in which computers learn patterns from data instead of being explicitly programmed with fixed rules. By using statistical and mathematical techniques, a machine learning model is trained on historical data and then uses the learned patterns to make predictions or decisions on new unseen data.

## 2: What are the main types of Machine Learning

There are three main types of machine learning: supervised learning, unsupervised learning, and reinforcement learning. In supervised learning, the model is trained on labeled data, meaning there is a target or output variable to predict. In unsupervised learning, the data is unlabeled, so there is no target variable, and the model tries to identify hidden patterns, groups, or structures within the data. In reinforcement learning, the model learns by interacting with an environment and receives rewards or penalties based on its actions, which helps improve its performance over time.


## 3: What is the difference between Supervised Learning and Unsupervised Learning?

In supervised learning, the model is trained on labeled data, whereas in unsupervised learning, the model is trained on unlabeled data. In supervised learning, there is a target variable to predict, so the model learns the relationship between the input features and the target output in order to make predictions on new unseen data. In unsupervised learning, there is no target variable to predict, so the model tries to identify hidden patterns, structures, or groups within the data.


## 4: What is the difference between Classification and Regression?


The difference between classification and regression lies in the type of target variable being predicted. In classification, the target variable is categorical, meaning the model predicts a class or category. In regression, the target variable is a continuous numerical variable, meaning the model predicts a numeric value. For example, predicting whether a passenger survived or not in the Titanic dataset is a classification problem, whereas predicting the price of a house is a regression problem.



## 5: What is a feature in Machine Learning?

A feature in machine learning refers to the input column or input variables used to train the machine learning model so that it can make predictions. It represents the characteristics or attributes of the data. For example, in a Titanic dataset, columns such as Age, Fare, Passenger Class can be used as features to predict whether a passenger survived.

## 6: What is label or target in Machine Learning?

A label or target variable is the output variable that the machine learning model tries to predict. In other words, it is the variable whose value we want the model to learn and predict based on the input features. For example, in the Titanic dataset, the Survived column is the target variable because the model predicts whether a passenger survived or not.

## 7: What is a training dataset and a testing dataset? Why is data splitting important?

A training dataset is the portion of the data used to train the machine learning model, meaning the model learns patterns and relationships from this data. A testing dataset is the portion of the data used to evaluate the performance of the trained model, and this data is not shown to the model during training. Data splitting is important because we need one part of the data to train the model and another unseen part to test how well the model performs on new data. This helps us check whether the model has actually learned the pattern or has just memorized the training data.

## 8: What is feature scaling and why is it needed in some algorithms?
Feature scaling is the process of bringing numerical input features to a common scale so that one feature does not dominate the others because of its larger magnitude. In machine learning datasets, different features can have very different ranges—for example, one feature may range from 1 to 100 while another ranges from 10,000 to 50,000. In such cases, some algorithms may give more importance to the feature with larger values, not because it is actually more important, but simply because of its scale. Therefore, feature scaling is needed to ensure that all features contribute more fairly to the model and to improve the performance of algorithms that are sensitive to feature magnitude, such as K-Nearest Neighbors (KNN), Support Vector Machines (SVM), Logistic Regression, K-Means Clustering, and Principal Component Analysis (PCA). On the other hand, tree-based algorithms such as Decision Trees and Random Forest are generally less affected by feature scaling.


## 9: What are Overfitting and Underfitting?

### Overfitting, Underfitting, Bias and Variance

An ideal machine learning model should have low bias and low variance. Such a model performs well on both training data and unseen test data, resulting in high training accuracy and high test accuracy with only a small difference between them.

#### Overfitting

Overfitting occurs when a model learns the training data too well and memorizes specific patterns and noise present in the training data. As a result, it achieves very high training accuracy but performs poorly on unseen test data.

An overfitted model has low bias because it can fit the training data very well. It has high variance because its performance on unseen data may vary significantly depending on how similar the test data is to the training data.

#### Underfitting

Underfitting occurs when a model is too simple to learn the underlying patterns in the data. As a result, it performs poorly on both training data and test data.

An underfitted model has high bias because it cannot learn the training data properly. It has low variance because its performance remains consistently poor even when evaluated on different unseen datasets.

#### Bias-Variance Trade-off

The bias-variance trade-off is the balance between underfitting and overfitting. A model with very high bias tends to underfit, while a model with very high variance tends to overfit. The goal is to build a model with low bias and low variance so that it generalizes well to unseen data.


### Overfitting, Underfitting, Bias and Variance (Student Analogy)

An ideal machine learning model should have low bias and low variance. Such a model performs well on both training data and unseen test data, resulting in high training accuracy and high test accuracy with only a small difference between them.

#### Underfitting

Consider a student who studies very little and does not understand the concepts properly.

Practice Test Marks:

* Test 1 = 40%
* Test 2 = 42%
* Test 3 = 38%

Final Exam Marks:

* Exam 1 = 41%
* Exam 2 = 39%
* Exam 3 = 43%

The student performs poorly in both practice tests and final exams. The marks remain consistently low regardless of the questions asked.

Similarly, an underfitted model performs poorly on both training and test data.

Characteristics:

* High Bias
* Low Variance
* Low Training Accuracy
* Low Test Accuracy

#### Overfitting

Consider a student who memorizes all the practice questions and answers without understanding the concepts.

Practice Test Marks:

* Test 1 = 100%
* Test 2 = 100%
* Test 3 = 99%

Final Exam Marks:

* Exam 1 = 85%
* Exam 2 = 70%
* Exam 3 = 60%

The student performs extremely well on the practice tests but the final exam marks vary significantly depending on how similar the exam questions are to the memorized questions.

Similarly, an overfitted model achieves very high training accuracy but its performance on unseen data may vary significantly.

Characteristics:

* Low Bias
* High Variance
* High Training Accuracy
* Low or Unstable Test Accuracy

#### Ideal Model

Consider a student who understands the concepts instead of memorizing.

Practice Test Marks:

* Test 1 = 95%
* Test 2 = 96%
* Test 3 = 94%

Final Exam Marks:

* Exam 1 = 93%
* Exam 2 = 92%
* Exam 3 = 94%

The student performs well on both practice tests and final exams because the concepts are understood properly.

Similarly, an ideal machine learning model learns the true patterns in the data and generalizes well to unseen data.

Characteristics:

* Low Bias
* Low Variance
* High Training Accuracy
* High Test Accuracy

#### Bias-Variance Trade-off

The bias-variance trade-off is the balance between underfitting and overfitting.

* Very High Bias → Underfitting
* Very High Variance → Overfitting
* Low Bias + Low Variance → Good Generalization

The goal is to build a model that performs well on both training and unseen data.


## 10. How Does a Linear Regression Model Work?

Linear Regression is a supervised machine learning algorithm used to predict continuous numerical values.

It works by finding the best-fit straight line that represents the relationship between the input variable(s) and the target variable.

The equation of a simple linear regression model is:

y = mx + c

Where:

* y = predicted value
* x = input feature
* m = slope or weight
* c = intercept

Initially, the model assigns random values to the weights and intercept. It then makes predictions and calculates the error between the predicted values and the actual values.

Using the Gradient Descent algorithm, the model continuously adjusts the weights and intercept to reduce the error. This process is repeated until the model finds the best-fit line with the minimum possible error.

Once trained, the model can use the learned relationship between the input and output variables to predict values for unseen data.
