## Handle Missing Values

## dropna()

```python
df_dropna = df.copy()

df_dropna = df_dropna.dropna()

df_dropna.shape
```

## fillna()

```python
df_fillna_age = df.copy()

df_fillna_age=df_fillna_age.fillna(df_fillna_age['Age'].mean()) # Mean imputation

df_fillna_age['Age'].isnull().sum()
```

```python
df_fillna_embarked = df.copy()

df_fillna_embarked = df_fillna_embarked.fillna(df_fillna_embarked['Embarked'].mode()[0]) # Mode imputation

df_fillna_embarked['Embarked'].isnull().sum()
```

reset_index() is used to restore the default row numbering (0, 1, 2, 3, ...) of a DataFrame. After operations such as deleting rows, filtering, or sorting, the row indices may become discontinuous. reset_index(drop=True) creates a new continuous index and removes the old index instead of keeping it as a separate column.
