## Subplots


```python
fig,axs = plt.subplots(2,1) ## 2rows and 1 column

axs[0].plot(x,np.sin(x))
axs[0].set_title("Sin Wave")

axs[1].plot(x,np.cos(x))
axs[1].set_title("Cos Wave")

plt.suptitle("Subplots Example")

plt.show()
```

plt.subplots(2, 1) creates a figure with multiple subplots arranged in 2 rows and 1 column. Here, fig represents the whole figure, and axs contains the individual subplot axes. axs[0] refers to the first subplot and axs[1] refers to the second subplot.


In subplots, the first number indicates the number of rows and the second number indicates the number of columns. If there are multiple rows and columns, subplots can be accessed using row and column positions such as axs[0,1], which refers to the first row and second column.

### Violin Plot

```python
sns.violinplot(x='class',y='age',data =df)

plt.title("Violin Plot - Age by Class")

plt.show()
```

### Interpretation of the Violin Plot

This violin plot is used to visualize the distribution of the numerical variable **Age** across the categorical variable **Passenger Class**. In this plot, the x-axis represents the different passenger classes , which are categorical groups, and the y-axis represents the numerical variable **Age**. Therefore, a violin plot is mainly used when we want to study the distribution of a **numerical variable across different categories**

The main purpose of a violin plot is to show not only the spread of the data but also the **density or concentration of observations** at different values. The width of the violin at any particular age indicates how densely the data points are concentrated at that age range. A wider region means more passengers belong to that age range, while a narrower region means fewer passengers are present there. Thus, violin plots are useful for understanding the overall shape of the distribution, where the values are concentrated, how widely the data is spread, and how the distribution differs between categories.

From this violin plot, we can compare the age distribution among first, second, and third class passengers. It appears that the age distribution of **First Class** passengers is spread over a wider range and has more concentration in the adult age group, especially around the 30–45 age range. **Second Class** passengers also show a concentration around the adult age group, but the distribution appears slightly narrower. **Third Class** passengers seem to have a concentration around younger adult ages, mostly in the 20–35 age range, with a somewhat narrower spread compared to First Class. So, this plot helps us understand that the age distribution is not exactly the same across passenger classes.

A violin plot is similar to a box plot in the sense that both are used to study the distribution of a numerical variable across categories. However, a **box plot** mainly shows the median, quartiles, spread, and possible outliers, whereas a **violin plot** provides additional information about the **shape and density of the distribution**. In other words, the box plot summarizes the data distribution, while the violin plot gives a richer picture of how the data values are distributed within each category.

### Swarm plot

```python
sns.swarmplot(x='class',y='age',data=df)

plt.title("Swarm Plot")

plt.show()
```

### Interpretation of the Swarm Plot

A swarm plot is used to display the individual observations of a numerical variable across different categories. In this plot, the categorical variable is `class` and the numerical variable is `age`. Each dot represents the age of a single passenger, and the points are arranged in such a way that they do not overlap with each other. This makes it easier to see the actual distribution of the data points, the spread of values, clusters of observations, and possible outliers within each passenger class. Unlike box plots or violin plots, which summarize the data, a swarm plot shows the actual underlying observations. Therefore, it is useful when we want to inspect how the real data points are distributed within each category.


### Strip Plot

```python
sns.stripplot(x='class',y='age',data=df,jitter=True)

plt.title("Strip Plot- Age by Class")

plt.show()
```

### Interpretation of the Strip Plot

A strip plot is used to display individual observations of a numerical variable across categories. In this plot, **class** is the categorical variable and  **age** is the numerical variable. Each dot represents the age of one passenger. The **jitter=True** parameter adds a small horizontal displacement to the points so that overlapping observations become more visible. This helps in understanding the spread and concentration of the data points within each passenger class. A strip plot is useful for visualizing the raw observations directly, although points may still overlap more than in a swarm plot.


### Swarm Plot vs Strip Plot

Both swarm plot and strip plot are used to visualize the individual observations of a numerical variable across categories. The main difference is that a swarm plot arranges the points carefully to avoid overlap, whereas a strip plot uses a small random horizontal displacement called jitter to reduce overlap. Therefore, swarm plots generally provide a clearer view of individual observations, while strip plots are simpler and faster for larger datasets.
