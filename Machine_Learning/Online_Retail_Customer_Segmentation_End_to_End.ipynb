## Preprocessing

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df['Description'] = df.groupby('StockCode')['Description'].transform( lambda x : x.ffill().bfill())
```

```python
df.isnull().sum()
```

```python
df.dropna(subset=['CustomerID','Description'],inplace=True)
```

```python
df = df[(df['Quantity']>0) & (df['UnitPrice']>0)]
```

```python
numerical_columns = [column for column in df.columns if df[column].dtype in ["int64", "float64"]]
print(numerical_columns)
```

```python
categorical_columns = [column for column in df.columns if df[column].dtype == "object"]
print(categorical_columns)
```

```python
datetime_columns = [column for column in df.columns if df[column].dtype == "datetime64[ns]"]
print(datetime_columns)
```

## EDA

### 1. Univariate Analysis

### Histogram

```python
for column in numerical_columns:
    plt.figure(figsize=(6,4))
    sns.histplot(df[column], kde=True, bins=30)
    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.show()

```

### Boxplot

```python
for column in numerical_columns:
    plt.figure(figsize=(6,4))
    sns.boxplot(x=df[column])
    plt.title(f"Box Plot of {column}")
    plt.xlabel(column)
    plt.show()
```

### Value Counts for Country

```python
df["Country"].value_counts()
```

### Count Plot 

```python
plt.figure(figsize=(12,6))
sns.countplot(
    y="Country",
    data=df,
    order=df["Country"].value_counts().index
)
plt.title("Count Plot of Country")
plt.xlabel("Count")
plt.ylabel("Country")
plt.show()
```

### Datetime Analysis

```python
print("Minimum Date :", df["InvoiceDate"].min())
print("Maximum Date :", df["InvoiceDate"].max())
```

```python

plt.figure(figsize=(8,4))
df["InvoiceDate"].dt.month.value_counts().sort_index().plot(kind="bar")
plt.title("Monthly Transactions")
plt.xlabel("Month")
plt.ylabel("Number of Transactions")
plt.show()
```

### Top 10 Countries

```python

top10 = df["Country"].value_counts().head(10)
```

```python

plt.figure(figsize=(10,5))
sns.barplot(x=top10.values, y=top10.index)
plt.title("Top 10 Countries by Number of Transactions")
plt.xlabel("Transactions")
plt.ylabel("Country")
plt.show()
```

## 2. Bivariate Analysis

### Scatter Plot

```python

plt.figure(figsize=(8,6))
sns.scatterplot(data=df, x="Quantity", y="UnitPrice", alpha=0.5)
plt.title("Quantity vs Unit Price")
plt.xlabel("Quantity")
plt.ylabel("Unit Price")
plt.show()
```

### Boxplot

```python


top10 = df["Country"].value_counts().head(10).index
```

```python


plt.figure(figsize=(12,6))
sns.boxplot(
    data=df[df["Country"].isin(top10)],
    x="Country",
    y="UnitPrice"
)
plt.xticks(rotation=45)
plt.title("Unit Price Distribution Across Top 10 Countries")
plt.show()
```

```python



plt.figure(figsize=(12,6))
sns.boxplot(
    data=df[df["Country"].isin(top10)],
    x="Country",
    y="Quantity"
)
plt.xticks(rotation=45)
plt.title("Quantity Distribution Across Top 10 Countries")
plt.show()
```

### Bar Plot

```python

top10_products = df["Description"].value_counts().head(10)

plt.figure(figsize=(12,6))
sns.barplot(x=top10_products.values, y=top10_products.index)
plt.title("Top 10 Most Frequently Purchased Products")
plt.xlabel("Number of Transactions")
plt.ylabel("Product")
plt.show()
```

### Line Plot

```python




monthly_transactions = df.groupby(df["InvoiceDate"].dt.month).size()

plt.figure(figsize=(8,5))
sns.lineplot(
    x=monthly_transactions.index,
    y=monthly_transactions.values,
    marker="o"
)
plt.title("Monthly Transactions")
plt.xlabel("Month")
plt.ylabel("Transaction Count")
plt.show()
```

## Multivariate Analysis

### Correlation Heatmap

```python

plt.figure(figsize=(5,4))
sns.heatmap(
    df[['Quantity', 'UnitPrice']].corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)
plt.title("Correlation Heatmap")
plt.show()
```

### Pair Plot

```python
sns.pairplot(
    df[['Quantity', 'UnitPrice']],
    diag_kind='hist'
)
plt.show()
```
