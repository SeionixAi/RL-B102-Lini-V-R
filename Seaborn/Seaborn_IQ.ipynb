# Seaborn Interview Questions

## 1. What is Seaborn and How is it Different from Matplotlib?

Seaborn is an open-source statistical data visualization library built on top of Matplotlib.

It provides advanced visualization capabilities and attractive default styles with simpler syntax.

### Advantages of Seaborn

- Easier to create statistical visualizations.
- Attractive default themes.
- Simpler syntax.
- Better integration with Pandas DataFrames.

### Difference Between Seaborn and Matplotlib

Seaborn:
- Built on top of Matplotlib.
- Provides advanced statistical visualizations.
- Requires less code.
- Better default styling.

Matplotlib:
- More flexible and customizable.
- Used as the foundation for many Python visualization libraries.

## 2. What is a Statistical Plot?

A statistical plot is a visualization used to represent statistical characteristics of data.

### Information Provided

- Distribution of data.
- Spread of data.
- Correlation between variables.
- Presence of outliers.
- Trends and patterns.

### Importance

Statistical plots help understand data before performing analysis and building machine learning models.

## 3. Difference Between Histplot and Boxplot

### Histplot

- Shows the distribution of numerical data.
- Uses bins to group values.
- Displays frequency within each interval.
- Helps identify skewness and distribution patterns.

### Boxplot

- Summarizes data using quartiles.
- Shows median and spread of data.
- Helps identify outliers.
- Useful for comparing distributions.

### Summary

Histplot → Distribution and frequency

Boxplot → Spread, quartiles, and outliers

## 4. What is a Heatmap and Where is it Used?

A heatmap is a color-coded visualization used to represent values in a matrix format.

It is commonly used to visualize correlations between numerical variables.

### Uses

- Correlation analysis.
- Feature selection.
- Identifying relationships between variables.
- Exploratory Data Analysis (EDA).

### Correlation Values

- +1 → Perfect Positive Correlation
- 0 → No Correlation
- -1 → Perfect Negative Correlation

Heatmaps help identify strongly related features in a dataset.

## 5. Why is Seaborn Useful for Data Exploration in AI Projects?

Seaborn helps explore and understand data through advanced statistical visualizations.

### Benefits

- Visualizes distributions.
- Detects outliers.
- Analyzes correlations.
- Supports feature selection.
- Assists in feature engineering.
- Simplifies data exploration.

### Importance in AI

Seaborn helps data scientists understand datasets before model training, leading to better preprocessing and more accurate machine learning models.
