# 1.Write a Python program to count the number of consonants in a given string.

# # Method 1: Using isalpha()

````python
def count_consonants(string):
    count=0
    vowels="aeiouAEIOU"

    for char in string:
        if char.isalpha() and  char not in vowels:
            count+=1
    return count
    
count_consonants("Hello, How are You")
````

## Method 2: Using ASCII Values

````python
# ASCII Ranges:
# A-Z -> 65 to 90
# a-z -> 97 to 122

# Special Characters:
# [ = 91
# \ = 92
# ] = 93
# ^ = 94
# _ = 95
# ` = 96

# Characters within the ranges 65-90 and 97-122
# are considered English alphabets.



def count_consonants(string):
    vowels="aeiouAEIOU"
    count=0
    for char in string:
        ascii_value=ord(char)
        if ((62<=ascii_value<=90 ) or(97<=ascii_value<=122)) and char not in vowels:
            count+=1
    return count

count_consonants("Malayalam")

````

## Method 3: Character Range Comparison (DSA Approach)

````python
def count_consonants(string):
    count=0
    vowels="aeiouAEIOU"
    for char in string:
        if (("a"<=char<="z") or ("A"<=char<="Z")) and char not in vowels:
            count+=1
    return count

count_consonants("Consonant")
````

# 2. Write a Python program to find the first non-repeating character in a string.

## Method 1: Dictionary (Hash Map) Approach

````python
def non_repeat_char(string):
    frequency=dict()
    for char in string:
        if char not in frequency:
           frequency[char]=1
        else:
            frequency[char]+=1
    
    for char in string:
        if frequency[char]==1:
            return char
    return None

non_repeat_char("MalayalaM")
````

````python
non_repeat_char("swiss")
````

## Method 2: Built-in count Approach

````python
def non_repeat_char(string):
    for char in string:
        if string.count(char)==1:
            return char
    return None

non_repeat_char("swiss")
````

# 3.Write a Python program to remove all duplicate characters from a string.

## Method 1: String Traversal and Membership Check

````python
def remove_duplicates(string):
    result=""
    for char in string:
        if char not in result:
            result+=char
    return result

remove_duplicates("hello")
````

### Time Complexity: O(n²)
### Space Complexity: O(n)

## Method 2: Dictionary (Hash Map) Approach

````python
def remove_duplicate(string):
    seen={}
    result=""
    for char in string:
        if char not in seen:
            seen[char]=1
            result+=char
        else:
            seen[char]+=1
    return result

remove_duplicate("Dictionary")

````

### Time Complexity: O(n)
### Space Complexity: O(n)

## Method 3: Set Approach


````python
def remove_duplicates(string):
    seen=set()
    result=""
    for char in string:
        if char not in seen:
            seen.add(char)
            result+=char
    return result

remove_duplicates("Hello How Are You")
````

### Time Complexity: O(n)
### Space Complexity: O(n)

# Complexity Analysis - Final Note

## What is Time Complexity?

Time Complexity measures how the running time of an algorithm grows as the input size increases.

## What is Space Complexity?

Space Complexity measures how much extra memory an algorithm requires as the input size increases.

---

# Method 1: String Traversal and Membership Check Approach

## Time Complexity: O(n²)

### Reason

- The outer loop traverses the string once → O(n)
- The condition `char not in result` searches the result string → O(n)

Therefore:

```text
O(n) × O(n) = O(n²)
```

## Space Complexity: O(n)

### Reason

- A new result string is created.
- In the worst case, all characters are unique.
- The result string can grow up to size n.

### Insight

- Less memory is used.
- No additional data structure is required.
- Membership checking is slower because strings use linear search.

---

# Method 2: Dictionary (Hash Map) Approach

## Time Complexity: O(n)

### Reason

- String traversal → O(n)
- Dictionary lookup (`char not in seen`) → O(1)

Therefore:

```text
O(n) × O(1) = O(n)
```

## Space Complexity: O(n)

### Reason

- The dictionary stores unique characters.
- The result string stores unique characters.

```text
O(n) + O(n) = O(2n)
```

In Big-O notation:

```text
O(2n) = O(n)
```

### Insight

- Extra memory is used to store previously seen characters.
- Faster lookup reduces time complexity.
- This is called a Time-Space Tradeoff.

---

# Method 3: Set Approach

## Time Complexity: O(n)

### Reason

- String traversal → O(n)
- Set lookup (`char not in seen`) → O(1)

Therefore:

```text
O(n) × O(1) = O(n)
```

## Space Complexity: O(n)

### Reason

- The set stores unique characters.
- The result string stores unique characters.

```text
O(n) + O(n) = O(2n)
```

In Big-O notation:

```text
O(2n) = O(n)
```

### Insight

- Similar to the Dictionary approach.
- Uses extra memory for faster lookup.
- Useful when only unique values need to be tracked.

---

# Key Learning

## Time-Space Tradeoff

Using additional memory such as a Dictionary or Set allows faster lookup operations.

```text
Less Memory  → Slower Lookup

More Memory  → Faster Lookup
```
