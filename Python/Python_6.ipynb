# Patterns

# 1. Draw a right-angled triangle pattern using stars.

```python
def right_angled_triangle(n):

  for i in range(1,n+1):
    print(" * " * i )

right_angled_triangle(6)
```

 # 2.Draw an inverted right-angled triangle pattern.

```python
def inverted_right_angled_triangle(n):

  for i in range(n,-0,-1):
    print(" * " *i)

inverted_right_angled_triangle(5)
```

# 3.Create a Square Frame Pattern

```python
def square_frame_pattern(n):

  for i in range(n):
    for j in range(n):

      if i==0 or i==n-1 or j==0 or j==n-1:
        print("*",end=" ")
      else:
        print(" ",end=" ")

    print()

square_frame_pattern(6)

```

# Solid Pattern

```python
def solid_pattern(n):
  for i in range(n):
    for j in range(n):
      print("*",end=" ")
    print()

solid_pattern(6)
```
