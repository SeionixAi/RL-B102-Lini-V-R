# Python Basics Interview Questions

## 1. What is Python? Why is it popular in AI and Data Science?

Python is a high-level, interpreted, and general-purpose programming language.

### High-Level
- Uses human-readable syntax.
- Hides low-level hardware details.
- Easier to learn and develop applications.

### Interpreted
- Code is executed line by line by the Python interpreter.
- No separate compilation step is required.

### General-Purpose
- Can be used for web development, automation, data analysis, machine learning, artificial intelligence, and software development.

### Why is it Popular in AI and Data Science?
- Simple and readable syntax.
- Large community support.
- Powerful libraries such as NumPy, Pandas, Matplotlib, Scikit-learn, TensorFlow, and PyTorch.
- Supports rapid development and experimentation.

## 2. What are Variables and Data Types in Python?

Variables are names used to store data in memory.

A data type defines the kind of value stored in a variable and determines the operations that can be performed on it.

Common data types:
- int
- float
- str
- bool
- list
- tuple
- set
- dictionary

## 3. What is Dynamic Typing in Python?

Python is a dynamically typed language, which means the data type of a variable is determined automatically at runtime.

A variable can store different types of values during program execution without explicitly declaring its type.

## 4. What is Type Casting? Give Examples.

Type casting is the process of converting one data type into another.

### Examples
- int()
- float()
- str()
- bool()

Type casting helps ensure compatibility between different data types.

## 5. What is the Difference Between = and ==?

### =
- Assignment operator.
- Used to assign a value to a variable.

### ==
- Comparison operator.
- Used to check whether two values are equal.

## 6. What is the Difference Between is and ==?

### ==
- Compares the values of two objects.

### is
- Compares the memory locations of two objects.
- Checks whether two variables refer to the same object in memory.

## 7. Difference Between List, Tuple, Set, and Dictionary

### List
- Ordered collection.
- Mutable.
- Allows duplicates.

### Tuple
- Ordered collection.
- Immutable.
- Allows duplicates.

### Set
- Unordered collection.
- Does not allow duplicates.

### Dictionary

- Keys must be unique.

## 8. What is the Difference Between Mutable and Immutable Data Types?

### Mutable
Objects can be modified after creation.

Examples:
- list
- dictionary
- set

### Immutable
Objects cannot be modified after creation.

Examples:
- string
- tuple

- float

## 9. Explain if, elif, and else

if, elif, and else are conditional statements used for decision making in Python.

### if
- Executes when the specified condition is True.

### elif
- Checks additional conditions when the previous condition is False.

### else
- Executes when none of the specified conditions are True.

These statements help control the flow of program execution based on different conditions.

## 10. Difference Between for Loop and while Loop

### for Loop
- Used when the number of iterations is known.
- Commonly used to iterate through sequences such as lists, tuples, strings, and ranges.

### while Loop
- Used when the number of iterations is not known in advance.
- Continues execution as long as a specified condition remains True.

Both loops are used for repetitive execution of code.

## 11. Difference Between break, continue, and pass

### break
- Terminates the loop immediately.

### continue
- Skips the current iteration and moves to the next iteration.

### pass
- Null statement that performs no action.
- Used as a placeholder when code will be implemented later.
- Helps avoid syntax errors when a code block is required.

## 12. What is a Function? Why Do We Use Functions?

A function is a reusable block of code that performs a specific task.

### Advantages of Functions

- Improves code reusability.
- Reduces code duplication.
- Improves readability.
- Makes programs easier to maintain.
- Helps organize code into smaller logical units.

Functions can accept inputs, process data, and return outputs.

## 13. What is the Purpose of the return Statement?

The return statement terminates a function and sends a value back to the caller.

### Advantages

- Returns output from a function.
- The returned value can be stored and reused.
- Helps in modular programming.
- Terminates function execution once executed.

### Difference Between return and print

- print() displays output on the screen.
- return sends a value back to the caller.
- Returned values can be stored and reused later in the program.

## 14. What are Local and Global Variables?

### Local Variables

- Declared inside a function.
- Accessible only within that function.
- Exist only during function execution.

### Global Variables

- Declared outside functions.
- Accessible throughout the program.
- Can be used by multiple functions.

## 15. What is List Comprehension?

List comprehension is a concise and readable way to create lists in Python.

### Advantages

- Reduces code length.
- Improves readability.
- Provides a compact alternative to loops for creating lists.

## 16. What is a Lambda Function?

A lambda function is an anonymous function that can take multiple arguments but contains only a single expression.

### Features

- Defined using the lambda keyword.
- Does not require a function name.
- Useful for short and simple operations.
- Often used with functions such as map(), filter(), and sorted().

## 17. What is Recursion? Give a Simple Example.

Recursion is a programming technique in which a function calls itself to solve a problem.

### Features

- A recursive function calls itself.
- Must contain a base condition to stop execution.
- Without a base condition, recursion will continue indefinitely and may cause a stack overflow.

### Example

A factorial function can be implemented using recursion.

## 20. What is Exception Handling?

Exception handling is the mechanism used to handle runtime errors and prevent unexpected program termination.

Python provides the following blocks:

- try
- except
- else
- finally

### try
Contains code that may raise an exception.

### except
Handles the exception if an error occurs.

### else
Executes when no exception occurs.

### finally
Executes regardless of whether an exception occurs or not.

### Advantages

- Prevents program crashes.
- Improves program reliability.
- Allows graceful error handling.

## 21. What is Exception Hierarchy in Python?

Exception hierarchy is the classification of exceptions into different levels based on inheritance.

### Hierarchy

BaseException
- Exception
  - TypeError
  - ValueError
  - IndexError
  - KeyError
  - ZeroDivisionError
  - FileNotFoundError

The hierarchy helps Python identify and handle different types of exceptions efficiently.

## 22. Difference Between try-except and try-except-finally

### try-except

- Used to handle exceptions.
- The except block executes when an exception occurs.

### try-except-finally

- Used to handle exceptions and perform cleanup operations.
- The finally block always executes regardless of whether an exception occurs.

### Common Use Cases

- Closing files.
- Releasing resources.
- Database connection cleanup.

## 23. What is File Handling in Python? Name Different File Modes.

File handling is the process of reading data from a file or writing data to a file.

Python uses the open() function to perform file operations.

### Common File Modes

- r  → Read mode
- w  → Write mode
- a  → Append mode
- x  → Create a new file
- r+ → Read and Write mode
- w+ → Write and Read mode
- a+ → Append and Read mode

### Advantages

- Stores data permanently.
- Allows reading and writing external files.
- Useful for data processing and reporting.

## 24. What is the Difference Between read(), readline(), and readlines()?

### read()

- Reads the entire file content at once.
- Returns a single string.

### readline()

- Reads one line at a time.
- Returns a single line as a string.

### readlines()

- Reads all lines from the file.
- Returns a list of strings, where each element represents a line.

## 25. What is the Purpose of the with Statement?

The with statement is used for resource management.

When used with file handling, it automatically closes the file after the operation is completed.

### Advantages

- No need to manually close files.
- Improves code readability.
- Prevents resource leaks.
- Handles cleanup automatically.

## 26. What are Iterators in Python?

An iterator is an object that returns elements one at a time.

Iterators use the next() function to retrieve values sequentially.

### Features

- Processes data one element at a time.
- Maintains its current position.
- Raises StopIteration when no elements remain.

## 27. What are Generators and How Are They Different from Functions?

Generators are special Python functions that use the yield keyword to produce values one at a time instead of returning all values at once.

### Features

- Use the yield keyword.
- Follow lazy evaluation.
- Generate values only when needed.
- Maintain state between executions.
- Memory efficient.

### Difference Between Generators and Functions

Generator:
- Uses yield.
- Produces values one by one.
- Can pause and resume execution.

Function:
- Uses return.
- Returns all results at once.
- Execution ends after return.

## 28. What is *args and **kwargs?

### *args

- Used to pass a variable number of positional arguments to a function.
- Useful when the number of arguments is not known in advance.

### **kwargs

- Used to pass a variable number of keyword arguments to a function.
- Stores arguments as key-value pairs.

### Advantages

- Provides flexibility in function calls.
- Allows functions to handle varying numbers of arguments.

## 29. What are Docstrings and Why Are They Important?

Docstrings are strings written inside modules, classes, or functions to describe their purpose and functionality.

### Importance

- Improves code documentation.
- Helps developers understand code.
- Improves maintainability.
- Can be accessed using the help() function.
