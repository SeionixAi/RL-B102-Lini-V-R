# 1. Dynamic Typing

## Question
Assign an integer value to a variable,
then reassign a string value to the same variable.

Print the value and datatype at each step.

## Explanation
Python is dynamically typed, meaning the same variable
can store different types of values during runtime.


```python
# Assign integer value
x = 10

# Display value and datatype
print(x, type(x))

# Reassign string value
x = "Hello Python"

# Display updated value and datatype
print(x, type(x))
```

# 2. Mutable vs Immutable Data Types

## Question
Modify a list and attempt to modify a tuple.

## Explanation
Lists are mutable because elements can be changed.
Tuples are immutable and cannot be modified after creation.

```python
# Create list
my_list = [1, 2, 3]

# Modify first element
my_list[0] = 10

# Display updated list
print(my_list)

# Create tuple
my_tuple = (1, 2, 3)

# Tuples are immutable
# my_tuple[0] = 10

# Display tuple
print(my_tuple)
```

Use a global variable and a local variable inside a function.

## Explanation

# 4. Return Statement in a Function

## Question
Create a function that takes two numbers
and returns their sum.

## Explanation
The return statement sends the result back
to the function caller.

```python
# Function definition
def add(a, b):

    # Return sum
    return a + b

# Store returned value
result = add(5, 3)

# Display result
print(result)
```

# 5. List Comprehension

## Question
Create a list of squares from 1 to 10.

## Explanation
List comprehension provides a compact way
to create lists in Python.


```python
# Create list of squares
squares = [x * x for x in range(1, 11)]

# Display list
print(squares)
```

# 6. break, continue, and pass

## Question
Demonstrate break, continue, and pass statements.

## Explanation
- continue skips current iteration
- break exits loop
- pass acts as placeholder

```python
# Loop from 1 to 10
for i in range(1, 11):

    # Skip iteration when i is 3
    if i == 3:
        continue

    # Exit loop when i is 5
    if i == 5:
        pass

    # Placeholder statement
    if i == 7:
        break

    # Display value
    print(i)
```

# 7. Recursion

## Question
Create a recursive function to find factorial of a number.

## Explanation
Recursion occurs when a function calls itself.

```python
# Recursive factorial function
def factorial(n):

    # Base condition
    if n == 0:
        return 1

    # Recursive call
    return n * factorial(n - 1)

# Display factorial
print(factorial(5))
```

# 8. Lambda Function

## Question
Create a lambda function to calculate square of a number.

## Explanation
Lambda functions are small anonymous functions
written in a single line.

```python
# Lambda function for square
square = lambda x: x * x

# Display result
print(square(4))
```

# 9. is vs ==

## Question
Compare two lists using == and is.

## Explanation
- == compares values
- is compares memory locations

```python
# Create lists
a = [1, 2, 3]
b = [1, 2, 3]

# Compare values
print(a == b)

# Compare memory locations
print(a is b)
```

# 10. Docstrings

## Question
Create a function using a docstring.

## Explanation
Docstrings are used to describe the purpose of functions,
classes, or modules.

```python
# Function definition
def greet():

    """This function prints a greeting message"""

    print("Hello Python")

# Function call
greet()

# Display docstring
print(greet.__doc__)
```
