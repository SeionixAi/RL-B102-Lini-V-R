#  1. Write a Python program to check whether a string contains only digits.

## Method 1: Using isdigit() Approach

### Time Complexity

O(n)

### Space Complexity

O(1)



```python
def is_only_digits(string):

  for char in string:
    if not char.isdigit():
      return False



  return True

is_only_digits("123a345")
```

```python
is_only_digits("3456123")
```

## Method 2: ASCII Value Approach

### Time Complexity

O(n)

### Space Complexity

O(1)



```python
def is_only_digits(string):

  for char in string:

    ascii_value=ord(char)

    if not (48<=ascii_value<=57): #Digits have ASCII values from 48 to 57.
      return False

  return True

is_only_digits("12k456")
```

```python
is_only_digits("012745")
```

## Method 3: Character Range Comparison (DSA Approach)

### Time Complexity

O(n)

### Space Complexity

O(1)



```python
def is_only_digits(string):

  for char in string:

    if not ('0'<=char<='9'):
      return False

  return True

is_only_digits("2123")
```

```python
is_only_digits("3e123")
```

# 2. Find the Longest Word in a String

---

## Method 1: String Splitting and Comparison Approach

### Time Complexity

O(n)

### Space Complexity

O(n)


```python
def longest_word(string):

  words=string.split()

  longest_word=""

  for word in words:

    if len(word) > len(longest_word):

      longest_word=word

  return longest_word

longest_word("Hello, Good Morning")
```

## Method 2: Pure String Traversal Approach (DSA Approach)

### Time Complexity

O(n)

### Space Complexity

O(n)


```python
def longest_word(string):

  longest_word=""
  current_word=""

  for char in string:

    if char!=" ":
      current_word+=char

    else:

      if len(current_word) > len(longest_word):
        longest_word=current_word

      current_word=""

  if len(current_word)>len(longest_word):
    longest_word=current_word

  return longest_word

longest_word("Python is Powerful")
```

# 3. Write a Python program to capitalize the first and last character of each word in a string.

## Method 1: String Manipulation Approach

### Time Complexity
O(n)

### Space Complexity

O(n)

```python
def capitalize_ends(string):

  words=string.split()
  result=[]


  for word in words:

    if len(word)>1:

      new_word=word[0].upper() +word[1:-1] +word[-1].upper()
      result.append(new_word)


    else:

      result.append(word.upper())





  return " ".join(result)

capitalize_ends("hi how are you")

```

## Method 2: Pure String Traversal Approach (DSA Approach)

### Time Complexity

O(n)

### Space Complexity

O(n)

```python
def capitalize_ends(string):

  result=""
  current_word=""

  for char in string:

    if char!=" ":
      current_word+=char

    else:

      if len(current_word)==1:
        result+=current_word.upper()

      elif len(current_word)>1:
        result+=current_word[0].upper()+current_word[1:-1]+current_word[-1].upper()


      result+=" "

      current_word=""


  if len(current_word)==1:
    result+=current_word.upper()

  elif len(current_word)>1:
    result+=current_word[0].upper()+current_word[1:-1]+current_word[-1].upper()

  return result

capitalize_ends("python is powerful")

```

# 4. Sum of Digits in a Number

## Method 1: String Conversion Approach

### Time Complexity

O(n)

### Space Complexity

O(n)



```python
def sum_of_digits(n):

  total=0
  for digit in str(n):
    total+=int(digit)

  return total

sum_of_digits(1234)
```

Time Complexity:
O(n)
Reason:
Traverse all digits in the string.

Space Complexity:
O(n)
Reason:
Creates a new string containing all digits.


## Method 2: Modulus and Division Approach (DSA Approach)

### Time Complexity

O(log n)

### Space Complexity

O(1)


```python
def sum_of_digits(n):

  total=0

  while n>0:

    digit=n%10

    total+=digit

    n=n//10

  return total

sum_of_digits(12345)
```

Time Complexity:
O(log n)
Reason:
Each iteration removes one digit.

Space Complexity:
O(1)
Reason:
Uses only fixed variables (total, digit, n).

# 5. Factorial of a Number

## Method 1: Iterative For Loop Approach

### Time Complexity

O(n)

### Space Complexity

O(1)



```python
def factorial(n):

  result=1

  for i in range(1,n+1):
    result*=i

  return result

factorial(5)
```

- Loop runs n times.
- Uses only one variable (`result`).

## Method 2: Recursive Approach

### Time Complexity

O(n)

### Space Complexity

O(n)



```python
def factorial(n):
  if n==0 or n==1:
    return 1

  return n*factorial(n-1)

factorial(5)
```

- Function is called n times.
- Each call is stored in the call stack.
- Therefore space complexity becomes O(n).

# 6. Fibonacci Sequence

## Method 1 : Iterative Approach

```python
def fibonacci_sequence(n):

  a,b=0,1

  for i in range(n):
    print(a,end=" ")
    a,b=b,a+b

fibonacci_sequence(6)
```

# 7.Find the Largest and Smallest Number in a List

```python
def smallest_largest(numbers):
  if not numbers:
    return None

  smallest=numbers[0]
  largest=numbers[0]

  for num in numbers:

    if num<smallest:
      smallest=num

    if num>largest:
      largest=num


  return smallest,largest

smallest_largest([3,0,-5,7,-20,0,45,-9])
```

Time Complexity: O(n)

Reason:
Traverse the list once.

Space Complexity: O(1)

Reason:
Only smallest and largest variables are used.
