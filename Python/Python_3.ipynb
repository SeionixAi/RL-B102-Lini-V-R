
# 1. Write a Python program to count the number of vowels in a given string.


```python
def count_vowels(text):

  vowels="aeiou"
  count=0

  for char in text.lower():
    if char in vowels:
      count+=1
  return count

print(count_vowels("Artificial Intelligence"))
```

# 2. Write a Python program to check whether a given string is a palindrome.

```python
def is_palindrome(text):

  #convert the inpu text into lowercase
  text_lower=text.lower()

  #reverse the lowercase text
  reversed_text=text_lower[::-1]

  #compare reversed_text and text_lower
  if reversed_text==text_lower:
    return True

  else:
    return False
```

```python
result1 = is_palindrome("madam")
print(result1)
```

```python
result2 = is_palindrome("python")
print(result2)
```

# 3. Write a Python program to check whether two strings are anagrams.



```python
def is_anagram(str1, str2):

    #Convert both strings to lowercase
    string1 = str1.lower()
    string2 = str2.lower()

    #Sort the characters of both strings
    sorted_string1 = sorted(string1)
    sorted_string2 = sorted(string2)

      #Compare the sorted character lists
    if sorted_string1 == sorted_string2:
        return True
    else:
        return False
```

```python
result1 = is_anagram("listen", "silent")
print(result1)
```

```python
result2 = is_anagram("hello", "world")
print(result2)
```
