# NumPy Interview Questions

## 1. What is NumPy?

NumPy (Numerical Python) is an open-source Python library used for numerical and scientific computing.

It provides the powerful ndarray data structure for storing and manipulating homogeneous data efficiently.

### Features

- Supports multidimensional arrays.
- Provides mathematical and statistical functions.
- Supports linear algebra operations.
- Enables vectorized computations.
- Widely used in Data Science, Machine Learning, and Artificial Intelligence.

## 2. What is an ndarray?

An ndarray (N-dimensional array) is the primary data structure provided by NumPy.

It stores homogeneous data and supports efficient numerical operations.

### Features

- Stores elements of the same data type.
- Supports one-dimensional, two-dimensional, and multidimensional data.
- Uses contiguous memory allocation.
- Optimized for high-performance computations.

## 3. Difference Between Array and List

### NumPy Array

- Stores homogeneous data.
- Uses contiguous memory allocation.
- Supports vectorized operations.
- Faster numerical computations.
- More memory efficient.

### Python List

- Can store heterogeneous data.
- Stores references to objects.
- Uses non-contiguous memory locations.
- Slower for numerical computations.
- Higher memory overhead.

## 4. Why is NumPy Faster Than Python Lists?

NumPy is faster than Python lists because it is implemented using optimized C code.

### Reasons

- Uses contiguous memory allocation.
- Supports vectorized operations.
- Reduces memory overhead.
- Performs operations at a lower level using optimized C implementations.
- Efficient for large-scale numerical computations.

## 5. What is Vectorization in NumPy?

Vectorization is the process of performing operations on an entire NumPy array at once without using explicit Python loops.

### Advantages

- Faster execution.
- Reduced Python overhead.
- Cleaner and more readable code.
- Efficient numerical computations.

Vectorization is one of the main reasons NumPy is faster than Python lists for mathematical operations.

## 6. What is Broadcasting?

Broadcasting is a NumPy mechanism that automatically expands compatible arrays or scalar values so that mathematical operations can be performed efficiently without explicitly copying data.

### Advantages

- Reduces the need for explicit loops.
- Improves performance.
- Simplifies array operations.
- Supports operations between arrays of different but compatible shapes.

## 7. What is the Difference Between reshape() and resize()?

### reshape()

- Changes the shape of an array.
- Does not change the total number of elements.
- Returns a reshaped view or copy of the array.

### resize()

- Changes the shape and size of an array.
- Can add or remove elements if necessary.
- Modifies the original array.

### Summary

reshape() changes only the shape, whereas resize() can change both the shape and the size of the array.

## 8. What are NumPy Dimensions and Axes?

Dimensions represent the number of axes in a NumPy array.

### Examples

- 1-D Array → 1 Dimension
- 2-D Array → 2 Dimensions
- 3-D Array → 3 Dimensions

### Axes

Axes represent the directions along which data is organized.

For a 2-D array:

- Axis 0 represents rows.
- Axis 1 represents columns.

Understanding dimensions and axes is important for performing operations on multidimensional arrays.

## 9. What is Slicing in NumPy Arrays?

Slicing is a technique used to extract a portion of an array using index ranges.

### Advantages

- Accesses subsets of data efficiently.
- Does not require explicit loops.
- Improves code readability.
- Useful for data manipulation and analysis.

Slicing allows selective retrieval of elements without modifying the original array.

## 10. How Does NumPy Help in Mathematical Computations for AI?

NumPy provides efficient numerical and mathematical operations that are widely used in Artificial Intelligence and Machine Learning.

### Features

- Supports vectorized computations.
- Provides matrix operations.
- Supports linear algebra functions.
- Enables efficient handling of large datasets.
- Optimized for high-performance numerical computing.

### Importance in AI

AI and Machine Learning algorithms rely heavily on matrix operations, linear algebra, and numerical computations. NumPy provides the performance and functionality required to perform these operations efficiently.
