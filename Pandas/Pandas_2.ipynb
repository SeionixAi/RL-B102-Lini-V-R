# Dataset 1: Titanic-Like Passenger Dataset

## Purpose
Learn:
- data inspection
- filtering
- missing value handling
- grouping and aggregation

## Create Dataset

Generate a Titanic-like passenger dataset using Numpy

```python
# Fix random values for reproducibility
np.random.seed(1)

#Create dataset with 200 rows
rows=200
```

```python
#Create  DataFrame
df = pd.DataFrame({

 #Generate Sequential Passenger IDs
"Passenger Id": np.arange(1,rows+1),

 #Randomly select  passenger names
 "Name": np.random.choice(
     ["Mr. John", "Mrs. Anna", "Miss Emma", "Dr. Smith"],
     rows

     ),
 #Genearte Age Column with some missing values
 "Age": np.random.choice(

                        #Combine random ages with NaN values
                        np.append(

                                  #Generate 180 random ages between 1 and 79
                                  np.random.randint(1,80,180),

                                  #Add 20 missing values
                                   [np.nan]*20
                                 ),
                        #Select 200 rows randomly
                        rows
 ),


 #Randomly assign  gender values
 "Sex": np.random.choice(
     ["male","female"],
     rows
 ),


  #Randomly assign passenger classes
  "Pclass": np.random.choice(
      [1,2,3],
      rows
  ),


 #Generate random fare values between 10 and 500
 "Fare": np.round(

            #Generate random decimal values
            np.random.uniform(10,500,
                              rows),

            #Round values to 2 decimal places.
            2

 )

})
```

# View Data

```python
#Display first 5 rows of the dataset
df.head()
```

```python
#Display last 5 rows of the dataset
df.tail()
```

```python
#Display 3 random rows from dataset
df.sample(3)
```

# Inspect Structure

```python
#Display number of rows and columns
df.shape
```

```python
# Display all Column names
df.columns
```

```python
# Display dataset information and missing values
df.info()
```

```python
# Display statistical summary of numerical columns
df.describe()
```

# Handle Missing Data

```python
#Count missing values in each column
df.isnull().sum()
```

```python
#Replace missing age values with average age
df["Age"]=df["Age"].fillna(df["Age"].mean())
```

```python
df["Age"].isnull().sum()
```

# Filtering Rows

```python
# Display Passengers whose age is greater than 30
df[df["Age"]>30]
```

```python
# Display only female passengers
df[df["Sex"]=="female"]
```

# Grouping & Aggregation

```python
#Calculate average fare for each passenger class
df.groupby("Pclass")["Fare"].mean()
```

# Dataset 2: Student Performance Dataset

## Purpose
Learn:
- sorting
- value counts
- apply()
- conditional columns
- grouping

## Create Dataset

Generate a student performance dataset using NumPy.

```python
np.random.seed(2)
rows = 150
```

```python
df = pd.DataFrame({
    "StudentId": np.arange(1, rows + 1),
    "Age": np.random.randint(18, 30, rows),
    "Gender": np.random.choice(["male", "female"], rows),
    "Course": np.random.choice(["AI", "DS", "ML", "Python"], rows),
    "Marks": np.random.randint(35, 100, rows)
})
```

```python
df.head()
```

# Sorting Data




```python
# Sort students based marks in descending order
df.sort_values(by="Marks",ascending=False)
```

# Value Counts

```python
#Count number of male and female students.
df["Gender"].value_counts()
```

# Apply Function

Create a Result column based on marks.
Students with marks greater than or equal to 50 are marked as Pass.

```python
# Create Result column using apply and lambda
df["Result"]=df["Marks"].apply(lambda x:"Pass" if x>=50 else "Fail")
```

```python
df["Result"].sample(10)
```

# Grouping

```python
# Calculate average marks for each course
df.groupby("Course")["Marks"].mean()
```

# Dataset 3: Employee Dataset

## Purpose
Learn:
- renaming columns
- dropping columns
- aggregation functions
- grouping operations

## Create Dataset

Generate an employee dataset using NumPy.

```python
np.random.seed(3)
rows = 120
```

```python
df = pd.DataFrame({
    "EmpId": np.arange(1, rows + 1),
    "Dept": np.random.choice(["HR", "IT", "Sales"], rows),
    "Salary": np.random.randint(30000, 150000, rows),
    "Gender": np.random.choice(["male", "female"], rows)
})

```

```python
df.head()
```

# Rename Columns


```python
#Reanme Dept column to Department
df.rename(columns={"Dept":"Department"},inplace=True)
```

```python
df.head()
```

# Drop Columns




```python
#Remove Gender column
df.drop("Gender",axis=1,inplace=True)
```

```python
df.head()
```

# Aggregation

```python
df.groupby("Department")["Salary"].agg("mean","max")
```
