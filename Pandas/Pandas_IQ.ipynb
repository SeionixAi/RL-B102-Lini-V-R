# Pandas Interview Questions

## 1. What is Pandas?

Pandas is an open-source Python library used for data manipulation, analysis, and preprocessing.

It provides powerful data structures such as Series and DataFrame for handling structured data efficiently.

### Features

- Data cleaning.
- Data transformation.
- Data analysis.
- Missing value handling.
- Data aggregation and grouping.

Pandas is widely used in Data Science, Machine Learning, and Artificial Intelligence projects.

## 2. What is a DataFrame?

A DataFrame is a two-dimensional labeled data structure consisting of rows and columns.

It is similar to a spreadsheet or database table and can store heterogeneous data types.

### Features

- Row and column labels.
- Supports multiple data types.
- Easy data manipulation and analysis.
- Can be created from files, dictionaries, lists, and databases.

## 3. Difference Between Series and DataFrame

### Series

- One-dimensional data structure.
- Contains a single column of data.
- Has an index and values.

### DataFrame

- Two-dimensional data structure.
- Contains rows and columns.
- Can store multiple Series.

A DataFrame can be considered a collection of Series.

## 4. What is CSV?

CSV stands for Comma-Separated Values.

It is a file format used to store tabular data in plain text form, where values are separated by commas.

### Advantages

- Simple and lightweight.
- Easy to share between applications.
- Widely supported by databases, spreadsheets, and programming languages.

## 7. What is the Difference Between loc and iloc?

### loc

- Label-based indexing.
- Retrieves data using row and column labels.
- Uses the actual names of rows and columns.

### iloc

- Integer position-based indexing.
- Retrieves data using row and column positions.
- Uses numerical indices.

### Summary

loc works with labels, whereas iloc works with integer positions.

## 8. What are Missing Values and How Does Pandas Handle Them?

Missing values are data points that are not available in a dataset and are typically represented as NaN (Not a Number).

### Methods to Handle Missing Values

- Mean Imputation
- Median Imputation
- Mode Imputation
- Forward Fill (ffill)
- Backward Fill (bfill)
- Removing Missing Values

### Importance

Handling missing values improves data quality and helps prevent inaccurate model predictions.
